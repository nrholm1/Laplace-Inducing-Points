```python
import jax
import jax.numpy as jnp
```

```python
def e(k, dim):
    """Return k'th std. basis vector in R^dim"""
    return jnp.zeros((dim,)).at[k].set(1.0)
    
def E(j, k, dim):
    """Return outer product of j'th and k'th std. basis vector in R^dim"""
    ej = e(j, dim)
    ek = e(k, dim)
    return jnp.outer(ej,ek)

D = 4
k = 3
theta = 0.22 * jnp.pi

I = jnp.eye(D)
ck = jnp.cos(theta)
sk = jnp.sin(theta)

I + (ck - 1) * (E(0,0,D) + E(k,k,D)) + sk * (E(0,k,D) - E(k,0,D))
```

```python
def givens(A, B):
    r = jnp.hypot(A, B)
    c = jnp.where(r != 0.0, A / r, 1.0)
    s = jnp.where(r != 0.0, B / r, 0.0)
    return r, c, s
```
